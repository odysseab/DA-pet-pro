### 3. Рассчитать стоимость хранения. 

Узнайте общий объем папки и сравните два числа: raw size (объем данных) и disk space consumed (с учетом репликации на BM серверах) у папки google_analytics

```python
!hdfs dfs -du -h -s /data/raw/google_analytics
```
